Check Missing Values — Markdown cell
## 10. Missing Value Analysis

Missing values can affect data analysis and machine learning.

The `isnull().sum()` function calculates the number of missing values in each column.

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

Missing Value Percentage — Code cell

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing value percentage:")
print(missing_percentage)
```

Display Only Columns with Missing Values — Code cell

```python
missing = df.isnull().sum()

missing = missing[missing > 0]

if len(missing) == 0:
    print("No missing values found in the dataset.")
else:
    print("Columns containing missing values:")
    print(missing)
```

## 11. Duplicate Row Analysis

Summary Table — Code cell

```python
summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Non-Null Count": df.notnull().sum().values,
    "Missing Count": df.isnull().sum().values,
    "Unique Values": df.nunique().values
})

summary
```

Automatic Dataset Observations
## 13. Basic Observations

The following section summarizes the main characteristics of the dataset.

```python
print("DATASET SUMMARY")

print(f"Number of rows    : {df.shape[0]}")
print(f"Number of columns : {df.shape[1]}")

print("\nFeatures:")
for column in df.columns:
    print("-", column)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
missing = df.isnull().sum()

for column, count in missing.items():
    if count > 0:
        print(f"- {column}: {count} missing value(s)")

if missing.sum() == 0:
    print("No missing values found.")

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nNumerical Columns:")
print(list(df.select_dtypes(include=np.number).columns))

print("\nCategorical Columns:")
print(list(df.select_dtypes(include=["object", "category"]).columns))
```
